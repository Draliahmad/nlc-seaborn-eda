```python
import pandas as pd
import os

# Ensure we are in the root directory
print("Current directory:", os.getcwd())

# Create the data folder if it does not exist
os.makedirs("data", exist_ok=True)

# Create the dataframe
df = pd.DataFrame({
    "Formulation": ["MPM-NLC","MPM-TO-NLC180","MPM-TO-NLC200","MPM-TO-NLC220","MPM-TO-NLC240"],
    "ThymeOil_mg": [0, 180, 200, 220, 240],
    "ParticleSize_nm": [51.35, 70.63, 72.42, 80.56, 88.73],
    "PDI": [0.269, 0.237, 0.202, 0.1518, 0.2518],
    "ZetaPotential_mV": [-32.53, -33.86, -33.99, -32.22, -33.22],
    "EncapsulationEfficiency_pct": [88.10, 88.39, 88.39, 91.93, 92.17]
})

# Save CSV
df.to_csv("data/nlc_data.csv", index=False)

print("CSV created successfully:")
print(os.listdir("data"))
df
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.set(style="whitegrid", context="talk")
```

```python
plt.figure(figsize=(9,6))

corr = df.drop(columns=["Formulation"]).corr()

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Correlation Heatmap of NLC Physicochemical Parameters")
plt.tight_layout()
plt.show()
```

```python
sns.pairplot(
    df.drop(columns=["Formulation"]),
    diag_kind="kde"
)
plt.show()
```

```python
plt.figure(figsize=(7,5))
sns.regplot(
    data=df,
    x="ParticleSize_nm",
    y="EncapsulationEfficiency_pct",
    scatter_kws={"s": 80}
)
plt.title("Particle Size vs Encapsulation Efficiency")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7,5))
sns.regplot(
    data=df,
    x="ZetaPotential_mV",
    y="EncapsulationEfficiency_pct",
    scatter_kws={"s": 80}
)
plt.title("Zeta Potential vs Encapsulation Efficiency")
plt.tight_layout()
plt.show()
```

# Nanomedicine Physicochemical Visualisation using Seaborn

## Overview
This project demonstrates exploratory data analysis (EDA) and data visualisation of **nanostructured lipid carrier (NLC)** formulations using Python and Seaborn. The aim is to illustrate how physicochemical parameters of nanomedicine formulations can be analysed and interpreted using modern data-science tools.

The dataset is derived from an MSc research project investigating **meropenem–thyme oil co-loaded NLCs** developed for respiratory infections.

---

## Dataset
The dataset includes physicochemical and formulation-related parameters commonly used in nanomedicine research:

- **Formulation** – NLC formulation code  
- **ThymeOil_mg** – Thyme oil concentration (mg)  
- **ParticleSize_nm** – Mean particle size (nm)  
- **PDI** – Polydispersity index  
- **ZetaPotential_mV** – Zeta potential (mV)  
- **EncapsulationEfficiency_pct** – Encapsulation efficiency (%)  

Data are stored in:


```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(9,6))
corr = df.drop(columns=["Formulation"]).corr()

sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap of NLC Physicochemical Parameters")
plt.tight_layout()
plt.savefig("figures/correlation_heatmap.png", dpi=300)
plt.show()
```

### Interpretation

The correlation heatmap highlights key relationships between formulation parameters.
A strong positive correlation is observed between thyme oil concentration and particle size,
indicating lipid matrix expansion with increasing oil content.
Encapsulation efficiency shows a strong positive association with particle size and a negative
association with PDI, suggesting that larger, more monodisperse NLCs favour improved drug entrapment.
These trends are consistent with reported behaviour of lipid-based nanocarriers.
